```python
num_mirrors = range(2,14,2)
maxQs =[]
maxQfreqs = []
vs =[]
for i in num_mirrors: 
    qi, fi, vi = u.defectandmirrors(500, length = 337, xlength= 218, ylengths = taper, mirrorlength=382, excitation = mp.Hz, nummirrors=i, findModes=True,
                                     modevisulization=False, smoothing=True)
    print(f'The max q is {max(qi)}, at a freq of {fi[qi.index(max(qi))]}, Mode Volume: {vi}')
    maxQs.append(max(qi))
    maxQfreqs.append(fi[qi.index(max(qi))])
    vs.append(vi)

plt.semilogy(num_mirrors, maxQs)
plt.semilogy(num_mirrors, np.array(maxQs)/np.array(vs))
```

```python
plt.semilogy(num_mirrors, maxQs)
plt.xlabel('Number of mirrors')
plt.ylabel('Q factor')
plt.title('Semilog Scale for Q versus number of mirrors')

print(vs)
Purcell = (3/(4*np.pi**2))*((1540e-9)/12)**3 *(np.array(maxQs)/np.array(vs))

```

```python
plt.plot(num_mirrors,Purcell)
plt.xlabel('Number of mirrors (on one side)')
plt.ylabel("Purcell Enhancement")
plt.title('Purcell Enhcancemnt versus mirrors')
```

```python
plt.plot(num_mirrors,vs)
plt.xlabel('num of mirrors')
plt.ylabel('Mode Volume')
plt.title('Mode Volume versus number of mirrors')
```
